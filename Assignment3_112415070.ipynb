```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])
n = len(x)
sum_x = np.sum(x)
sum_y = np.sum(y)
sum_xy = np.sum(x * y)
sum_x2 = np.sum(x * x)
m = (n * sum_xy - sum_x * sum_y)/(n * sum_x2 - sum_x ** 2)
c = (sum_y -m *sum_x)/n
print("Slope =",m)
print("Intercept =",c)
y_pred = m * x + c
print("Predicted values:",y_pred)
plt.scatter(x,y)
plt.plot(x,y_pred)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Least Square Approximation")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])
m = 0
c = 0
learning_rate = 0.01
iterations = 1000
n = len(x)
for i in range(iterations):
    y_pred = m * x +c
    error = y_pred -y
    dm = (2 /n) *np.sum(error *x)
    dc = (2 /n) *np.sum(error)
    m = m -learning_rate *dm
    c = c -learning_rate *dc

print("Slope =",m)
print("Intercept =",c)

y_pred = m * x +c

print("Predicted values:",y_pred)
plt.scatter(x,y)
plt.plot(x,y_pred)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression using Gradient Descent")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv("Housing.csv")
data = pd.get_dummies(
    data,
    columns=[
        "mainroad",
        "guestroom",
        "basement",
        "hotwaterheating",
        "airconditioning",
        "prefarea",
        "furnishingstatus"
    ],
    drop_first=True
)
X = data.drop("price", axis=1).values
y = data["price"].values
X = X.astype(float)
y = y.astype(float)
X = np.c_[np.ones(X.shape[0]),X]

theta = np.linalg.inv(X.T @ X) @ X.T @ y
y_pred = X @ theta
mse = np.mean((y - y_pred) **2)
print("Parameters:")
print(theta)
print("Mean Squared Error:",mse)
plt.scatter(y,y_pred)
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Housing Price Prediction")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv("Titanic-Dataset.csv")
data = data[["Pclass", "Age", "SibSp", "Parch", "Fare"]]
data = data.dropna()
X = data[["Pclass", "Age", "SibSp", "Parch"]].values
y = data["Fare"].values
X = np.c_[np.ones(X.shape[0]), X]
theta = np.linalg.inv(X.T @ X) @ X.T @ y
y_pred = X @ theta
mse = np.mean((y - y_pred) ** 2)
print("Parameters:")
print(theta)
print("Mean Squared Error:",mse)
plt.scatter(y,y_pred)
plt.xlabel("Actual Fare")
plt.ylabel("Predicted Fare")
plt.title("Titanic Linear Regression")
plt.show()
```
